````python
# --- Slide 17: Functions (Greet Example) ---
def greet(name):
    print(f"Hello, {name}!")
    return f"Hello, {name}!"

print(greet("Alice"))
print(greet("Bob"))
print(greet("Charlie"))


# --- Slide 18: Functions in Practice ---
def normalize(values):
    """Scale values to a 0-1 range."""
    low = min(values)
    high = max(values)
    return [(v - low) / (high - low) for v in values]

def average(values):
    return sum(values) / len(values)

grades = [88, 91, 76, 95, 84]
scaled = normalize(grades)
mean_grade = average(grades)

print(scaled)
print(mean_grade)
````
